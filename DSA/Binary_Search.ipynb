### Exact Binary Search
- [l, r]
- Does v exist? if so, where?

```python
# Check m first, and is not target? exclude it => l = m + 1, r = m - 1
nums = [num for num in range(1, 10)]

def exact_search(nums, v):
    l = 0
    r = len(nums) - 1
    while l <= r:
        m = (l + r) // 2
        if nums[m] == v:
            return (m, v)
        elif nums[m] < v:
            l = m + 1
        else:
            r = m - 1

    return -1

exact_search(nums, 7)
```

### Lower-bound Binary Search
- [l, r)
- First index whose value is greater than or equal to v
- Equivalent to bisect_left

```python
nums = [1, 3, 3, 4, 6, 7, 9]

def lower_bound_search(nums, v):
    l, r = 0, len(nums)

    while l < r:
        m = (l + r) // 2
        if nums[m] < v:
            l = m + 1   # Exclude m as value at m is less than v.
                        # We want m whose value is greater then or equal to v
        else:
            r = m       # nums[m] >= v, means m could be first valid position

    return (l, nums[l])

lower_bound_search(nums, 3)
```

### Upper-boundary Binary Search
- First index whose value is strictly greater than v
- Equivalent to bisect_right

```python
nums = [1, 3, 3, 4, 6, 7, 9]

def upper_bound_search(nums, v):
    l, r = 0, len(nums)

    while l < r:
        m = (l + r) // 2
        if nums[m] <= v:
            l = m + 1
        else:
            r = m

    return (l, nums[l])

upper_bound_search(nums, 3)
```
