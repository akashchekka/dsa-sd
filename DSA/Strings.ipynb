### Luhn's Algorithm - Credit Card number check

```python
string = '36101025079892'

str_array = [int(string[i]) for i in reversed(range(len(string)))]

sum = 0
for i, x in enumerate(str_array):
    if i & 1:
        doubled = (2 * x)
        doubled_mod_nine = doubled - 9 if doubled > 9 else doubled
        sum += doubled_mod_nine
    else:
        sum += x

sum % 10 == 0
```

### Minimum Window Substring

```python
# ADOBECODEBANC <==> BAC

from collections import defaultdict

def min_window_substring(s:str , t:str):
    window = defaultdict(int)
    need = defaultdict(int)

    for c in t:
        need[c] += 1

    valid = 0
    start, length = 0, float('inf')
    l = r = 0

    while r < len(s):
        current = s[r]
        r += 1

        # Update the valid if the count of char is equal to count in need
        if current in need:
            window[current] += 1
            if window[current] == need[current]:
                valid += 1

        while valid == len(need):
            if r - l < length:
                start = l
                length = r - l # Update length once valid = len(need) 

            l2 = s[l]
            l += 1

            # Shrink window from left
            if l2 in need:
                if window[l2] == need[l2]:
                    valid -= 1
                window[l2] -= 1
    
    if length == float('inf'): return ""

    return s[start:start+length]

min_window_substring("ADOBECODEBANC", "ABC")
```

### String to Int

```python
s = '  -042'

def str_to_int(s: str):
    s = s.lstrip()

    if not s:
        return 0
    
    i = 0
    sign = -1 if s[i] == '-' else 1
    i += 1 if s[i] in ['-', '+'] else 0

    res = 0
    while i < len(s) and s[i].isdigit():
        res = res * 10 + int(s[i])
        i += 1

    res *= sign

    return max(-2**31, min(res, 2**31 - 1))

str_to_int(s)
```

### Longest Character Replacement

```python
from collections import defaultdict

def longest_char_repl(s: str, k: int):
    l = maxLen = maxF = 0
    counter = defaultdict(int)

    for r in range(len(s)):
        counter[s[r]] += 1
        maxF = max(maxF, counter[s[r]])

        if (r - l + 1) - maxF > k:
            counter[s[l]] -= 1
            l += 1
        else:
            maxLen = max(maxLen, r - l + 1)

    return maxLen

longest_char_repl('AAABBABB', 2)
```

```python
def longestPalindrome(s: str) -> str:
    res = ""
    maxlen = 0

    def process(l, r):
        nonlocal res, maxlen
        while l >= 0 and r < len(s) and s[l] == s[r]:
            if r - l + 1 > maxlen:
                res = s[l:r+1]
                maxlen = r - l + 1
            l -= 1
            r += 1

    for i in range(len(s)):
        l, r = i, i
        process(l, r)

        l, r = i, i + 1
        process(l, r)

    return res

longestPalindrome('ababd')
```

### First Non-Repeating Character in a Stream of Data

```python
from collections import deque, defaultdict

def first_non_rep_char(s: str):
    hmap = defaultdict(int)
    q = deque()
    res = []

    for char in s:
        hmap[char] += 1
        q.append(char)

        while q and hmap[q[0]] > 1:
            q.popleft()
        
        if not q:
            res.append(-1)
        else:
            res.append(q[0])

    return res

first_non_rep_char('abab')
```

### Resolve file path

```python
def resolve_path(path: str) -> str:
    stack = []

    for part in path.split('/'):
        if part == '' or part == '.':
            continue          # ignore empty parts and current-dir
        elif part == '..' and stack:
            stack.pop()   # go up one directory
        else:
            stack.append(part)

    return '/' + '/'.join(stack)

resolve_path('/a/b/../c/./d')
# resolve_path('.')
```

### Permutation in string

```python
def permutation_in_string(s1: str, s2: str):
    if len(s1) > len(s2):
        return False

    s1_count, s2_count = [0] * 26, [0] * 26
    for i in range(len(s1)):
        s1_count[ord(s1[i]) - ord('a')] += 1
        s2_count[ord(s2[i]) - ord('a')] += 1

    matches = 0
    for i in range(26):
        matches += 1 if s1_count[i] == s2_count[i] else 0

    l = 0
    for r in range(len(s1), len(s2)):
        if matches == 26:
            return True

        idx = ord(s2[r]) - ord('a')
        s2_count[idx] += 1
        if s1_count[idx] == s2_count[idx]:
            matches += 1
        elif s1_count[idx] + 1 == s2_count[idx]:
            matches -= 1

        idx = ord(s2[l]) - ord('a')
        s2_count[idx] -= 1
        if s1_count[idx] == s2_count[idx]:
            matches += 1
        elif s1_count[idx] - 1 == s2_count[idx]:
            matches -= 1

        l += 1

    return matches == 26
```

### Encode and Decode Strings

```python
def encode(strs: list[str]):
    if not strs:
        return ''

    res = []
    for s in strs:
        res.append(str(len(s)))
        res.append('#')
        res.append(s)
    return ''.join(res)

def decode(s: str):
    res = []
    i, j = 0, 0
    while j < len(s):
        if s[j].isdigit():
            j += 1
            continue
        if s[j] == '#':
            num = int(s[i:j])
            res.append(s[j + 1: j + 1 + num])
            j = j + 1 + num
            i = j

    return res

s = encode(['abc', 'defg', 'hi'])
print(s)
decode(s)
```

### CSV Parser

```python
def csv_parser(text: str):
    n = len(text)
    csv = []
    field = []
    row = []
    i = 0
    quote = False

    while i < n:
        ch = text[i]
        if quote:
            if ch != '"':
                field.append(ch)
            else:
                if i + 1 < n and text[i+1] == '"':
                    field.append('"')
                    i += 1
                else:
                    quote = False
        else:
            if ch == ',':
                row.append(''.join(field))
                field = []
            elif ch == '\n':
                row.append(''.join(field))
                csv.append(row)
                row = []
                field = []
            elif ch == '"':
                quote = True
            elif ch == '\\':
                if i + 1 < n:
                    field.append(text[i + 1])
                i += 1
            else:
                field.append(ch)
        i += 1
    return csv

text = '''a,b,c
d,e,\'f
g,h,"i,""Main"" abc"
'''

csv_parser(text)
```
