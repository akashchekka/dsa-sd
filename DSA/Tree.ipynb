```python
class Node:
    def __init__(self, data: int):
        self.data = data
        self.left = None
        self.right = None
```

```python
import collections

class BST:
    def __init__(self):
        self.head = None

    ### Insert Nodes to tree
    def insert(self, data: int):
        self.head = self.add_node(self.head, data)

    def add_node(self, head: Node, data: int):
        if head is None:
            return Node(data)
        
        if data < head.data:
            head.left = self.add_node(head.left, data)
        elif data > head.data:
            head.right = self.add_node(head.right, data)

        return head
    
    ### Print tree
    def print_tree(self, head: Node):
        if head is None:
            return None
        
        self.print_tree(head.left)
        print(head.data, end=' ')
        self.print_tree(head.right)

    def display(self):
        self.print_tree(self.head)

    ### BFS
    def bfs(self, head: Node):
        queue = collections.deque()
        queue.append(head)

        while queue:
            first_node = queue.popleft()
            print(first_node.data)
            if first_node.left:
                queue.append(first_node.left)
            if first_node.right:
                queue.append(first_node.right)

    ### right view
    def right_view(self, head: Node):
        q = collections.deque()
        q.append(head)

        while q:
            q_len = len(q)
            rightNode = None
            for _ in range(q_len):
                node = q.popleft()
                if node:
                    rightNode = node
                    if node.left:
                        q.append(node.left)
                    if node.right:
                        q.append(node.right)
            print(rightNode.data, end=' ')

    ### Depth of tree:
    def max_depth(self, head: Node):
        def dfs(head):
            if head is None:
                return -1
            
            return max(self.max_depth(head.left), self.max_depth(head.right)) + 1

        def bfs(head):
            if not head:
                return -1

            q = collections.deque()
            q.append(head)

            depth = -1

            while q:
                depth += 1

                for _ in range(len(q)):
                    node = q.popleft()

                    if node.left:
                        q.append(node.left)
                    if node.right:
                        q.append(node.righ)

            return depth

    def depth_of_node(self, head: Node, val: int):
        def dfs1(head):
            if head is None:
                return -1

            if head.data == val:
                return 0

            if val < head.data:
                depth = dfs1(head.left)
            else:
                depth = dfs1(head.right)

            return -1 if depth == -1 else depth + 1

        def dfs2(head):
            depth = 0

            while head:
                if head.data == val:
                    return depth

                head = head.left if val < head.data else head.right
                depth += 1

            return -1
        
    ### Print level k
    def print_level_util(self, head: Node, c: int, k: int):
        if head is None:
            return
        
        if c == k:
            print(head.data, end=' ')
            self.levelCount += 1
            return

        self.print_level_util(head.left, c+1, k)
        self.print_level_util(head.right, c+1, k)

    def print_level(self, k: int):
        c = 0
        self.levelCount = 0
        self.print_level_util(self.head, c, k)
        print("\ncount: " + str(self.levelCount))

    def delete_key(self, key):
        def delete_key_util(self, node, key):
            if not node:
                return None
            
            temp = node

            if key < temp.data:
                temp.left = self.delete_key_util(temp.left, key)
            elif key > temp.data:
                temp.right = self.delete_key_util(temp.right, key)
            else:
                if temp.right is None:
                    return temp.left
                if temp.left is None:
                    return temp.right
                
                leafNode = temp.right
                while leafNode.left:
                    leafNode = leafNode.left
                
                temp.data = leafNode.data
                temp.right = self.delete_key_util(temp.right, leafNode.data)

            return temp
        
        return delete_key_util(self.head, key)

    def lowest_common_ancestor(self, p: Node, q: Node):
        node = self.head

        while node:
            if p.data > node.data and q.data > node.data:
                node = node.right
            elif p.data < node.data and q.data < node.data:
                node = node.left
            else:
                return node.data

    def kthSmallest(self, k):
        node = self.head

        c = 0
        res = 0

        def traverse(node):
            nonlocal c, res

            if not node:
                return None
            
            traverse(node.left)

            c += 1
            if c == k:
                res = node.data
                return

            traverse(node.right)

        traverse(node)
        print(res)

    def isSubtree(self, root, subRoot) -> bool:
        if not root:
            return False

        if not subRoot:
            return True

        def verify(root1, root2):
            if not root1 and not root2:
                return True
            
            if not root1 or not root2:
                return False

            if root1.val != root2.val:
                return False

            return (True and 
                    verify(root1.left, root2.left) and 
                    verify(root1.right, root2.right))

        def inorder(root):
            if not root:
                return False

            if root.val == subRoot.val:
                if verify(root, subRoot):
                    return True

            return inorder(root.left) or inorder(root.right)

        return inorder(root)

    def isValidBST(self, root: Node) -> bool:
        from collections import deque
        
        if not root:
            return True

        def bfs(node: Node, l: float, r: float):
            q = deque([(root, float('-inf'), float('inf'))])
            while q:
                curr, l, r = q.popleft()

                if not (l < curr.data < r):
                    return False

                if curr.left:
                    q.append((curr.left, l, curr.data))
                if curr.right:
                    q.append((curr.right, curr.data, r))
            return True
        
        def dfs(node: Node, l: float, r: float):
            if not node: return True
            if not (l < node.data < r): return False

            return (
                dfs(node.left, l, node.data) and
                dfs(node.right, node.data, r)
            )
        
        return dfs(root, float('-inf'), float('inf'))
    
    def serialize_and_deserialize(self, root: Node):
        def serialize(root: Node):
            res = []

            def dfs(node):
                if not node:
                    res.append('N')
                    return None
                
                res.append(str(root.val))
                dfs(root.left)
                dfs(root.right)
            
            dfs(root)
            return ','.join(res)
        
        def deserialize(data: str):
            data = data.split(',')
            i = 0

            def dfs():
                nonlocal i
                if data[i] == 'N':
                    i += 1
                    return None
                
                node = Node(int(data[i]))
                i += 1

                node.left = dfs()
                node.right = dfs()

                return node
            
            return dfs()
        
    def diameter(self, root:Node):
        res = 0

        def dfs(node):
            nonlocal res

            if not node:
                return 0
            
            left = dfs(node.left)
            right = dfs(node.right)
            
            res = max(res, left + right)

            return 1 + max(left, right)
        
        dfs(root)

        return res
    
    def invert_tree(self, root: Node):
        if not root:
            return None
        
        root.left, root.right = root.right, root.left

        self.invert_tree(root.left)
        self.invert_tree(root.right)

        return root
    
    def max_path(self, root: Node):
        res = float('-inf')

        def path_traverse(node: Node):
            nonlocal res

            if not node:
                return 0
            
            left = max(0, path_traverse(node.left))
            right = max(0, path_traverse(node.right))

            res = max(res, node.data + left + right)

            return node.data + max(left, right)
        
        path_traverse(root)

        return res
    
    def isBalanced(root):
        def dfs(node):
            if not node:
                return [True, 0]

            left, right = dfs(node.left), dfs(node.right)
            balanced = left[0] and right[0] and abs(left[1] - right[1]) <= 1

            return [balanced, 1 + max(left[1], right[1])]
        
        return dfs(root)[0]

    def goodNodes(self, root: Node):
        res = 0

        def dfs(node, prev_val):
            nonlocal res

            if not node:
                return

            if node.val >= prev_val:
                res += 1
            
            prev_val = max(prev_val, node.val)
            dfs(node.left, prev_val)
            dfs(node.right, prev_val)
        
        dfs(root, root.val)
        return res

    def top_view(root: Node):
        first_at_vertical = {}
        q = collections.deque([(root, 0)])
        while q:
            node, vertical = q.popleft()
            if vertical not in first_at_vertical:   # uncommenting this makes this bottom view
                first_at_vertical[vertical] = node.data

            if node.left:
                q.append((node.left, vertical - 1))
            if node.right:
                q.append((node.right, vertical + 1))

        return [first_at_vertical[v] for v in sorted(first_at_vertical)]
```

### Build Tree from Inorder and Preorder
- Inorder:      Left, Root, Right
- Preorder:     Root, Left, Right
- Postorder:    Left, Right, Root

```python
def build_in_pre(inorder: list[int], preorder: list[int]):
    in_map = {x: i for i, x in enumerate(inorder)}
    in_start, pre_start = 0, 0
    in_end, pre_end = len(inorder) - 1, len(preorder) - 1

    def construct(in_start, in_end, pre_start, pre_end):
        if (in_start > in_end) or (pre_start > pre_end):
            return None

        in_root = in_map[preorder[pre_start]] # Inorder root index
        remaining_left = in_root - in_start

        node = Node(preorder[pre_start])
        node.left = construct(in_start, in_root - 1, pre_start + 1, pre_start + remaining_left)
        node.right = construct(in_root + 1, in_end, pre_start + remaining_left + 1, pre_end)

        return node

    node = construct(in_start, in_end, pre_start, pre_end)
    return node
```

### Binary Tree from Preorder and Postorder

```python
def constructFromPrePost(preorder: list[int], postorder: list[int]):
    pre_map = {x: i for i, x in enumerate(preorder)}

    def helper(pre_start, pre_end, post):
        if pre_start > pre_end:
            return None
    
        node = Node(post.pop()) # Root node is at post[-1]
        if pre_start == pre_end:
            return node

        ind = pre_map[post[-1]] # Index of right child's root will be at post[-1] after above pop
    
        node.right = helper(ind, pre_end, post)
        node.left = helper(pre_start + 1, ind - 1, post)

        return node
    
    pre_start, pre_end = 0, len(preorder) - 1
    return helper(pre_start, pre_end, postorder)

def constructFromPrePost(preorder: list[int], postorder: list[int]):
    pre_map = {x: i for i, x in enumerate(preorder)}

    def helper(pre_start, pre_end, post_end):
        if pre_start > pre_end:
            return None
    
        node = Node(postorder[post_end]) # Root node is at post[-1]
        if pre_start == pre_end:
            return node

        ind = pre_map[postorder[post_end - 1]] # Index of right child's root in pre will be at post[-1] after above pop
    
        node.right = helper(ind, pre_end, post_end - 1)
        right_size = pre_end - ind + 1
        node.left = helper(pre_start + 1, ind - 1, post_end - right_size - 1)

        return node

    pre_start, pre_end = 0, len(preorder) - 1
    post_end = len(postorder) - 1
    return helper(pre_start, pre_end, post_end)
```
