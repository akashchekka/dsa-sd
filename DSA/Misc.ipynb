### FIrst Missing Positive Integer

```python
def missing(nums: list[int]):
    n = len(nums)

    # Convert all negatives to 0
    for i in range(n):
        if nums[i] < 0:
            nums[i] = 0

    # nums[abs(nums[i]) - 1] *= -1  
    for i in range(n):
        value = abs(nums[i])
        if 1 <= value <= n + 1:
            if nums[value - 1] > 0:
                nums[value - 1] *= -1
            elif nums[value - 1] == 0:
                nums[value - 1] = -1 * (n + 1)

    for i in range(1, n+1):
        if nums[i - 1] < 0:
            return i

    return n + 1        
```

### pow(x, n)

```python
def powx_n(x: float, n: int):
    if n == 0:
        return 1.0
    if n < 0:
        return 1/powx_n(x, -n)
    if n & 1:
        return x * powx_n(x, n - 1)
    else:
        return powx_n((x*x), n >> 1)

powx_n(2.0000, -2)
```
